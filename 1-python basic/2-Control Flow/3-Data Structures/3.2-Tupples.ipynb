##### Introduction to Tuples
Explanation:

Tuples are ordered collections of items that are immutable.
They are similar to lists, but their immutability makes them different.


```python
## Introduction to Tuples in Python
##Tuples are a built-in data structure in Python that are similar to lists but have some key differences. A tuple is an ordered collection of items that is immutable, meaning that once a tuple is created, its contents cannot be changed. Tuples are defined using parentheses `()` and can contain elements of different data types.
# Creating a tuple
my_tuple = (1, "hello", 3.14, [1, 2, 3]) # this is a tuple that contains an integer, a string, a float, and a list
print(my_tuple)  # Output: (1, 'hello', 3.14, [1, 2, 3])


## converting between list and tuple
lst = list(my_tuple)  # Convert tuple to list
print("lst:", lst)  # Output: [1, 'hello', 3.14, [1, 2, 3]] 
tpl = tuple(lst)  # Convert list back to tuple
print("tpl:", tpl)  # Output: (1, 'hello', 3.14, [1, 2, 3])

numbers=tuple([1,2,3,4,5,6]) # Convert a list of numbers to a tuple
numbers 
```

```python
## Accessing Tuple Elements
print("full Tuple:", numbers)
print("First element:", numbers[0])  # Output: 1 (first element)
print("Third element:", numbers[2])  # Output: 3 (third element)
print("Last element:", numbers[-1]) # Output: 6 (last element)
print("First four elements:", numbers[0:4])  # Output: (1, 2, 3, 4) (slicing the tuple to get the first four elements)
print("Reversed tuple:", numbers[::-1]) # Output: (6, 5, 4, 3, 2, 1) (reversing the tuple using slicing)

```

```python
## Tuple Operations
# Concatenation
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated_tuple = tuple1 + tuple2
print("Concatenated Tuple:", concatenated_tuple)  # Output: (1, 2, 3, 4, 5, 6)
# Repetition
repeated_tuple = tuple1 * 3
print("Repeated Tuple:", repeated_tuple)  # Output: (1, 2, 3, 1, 2, 3, 1, 2, 3)
# Membership Testing
print(2 in tuple1)  # Output: True (2 is in tuple1)
print(5 in tuple1)  # Output: False (5 is not in tuple1)
# Length of a Tuple
print("Length of tuple1:", len(tuple1))  # Output: 3 (number of elements in tuple1) 
```

```python
## Immutable Nature Of Tuples
## Tuples are immutable, meaning their elements cannot be changed once assigned.
my_tuple = (1, 2, 3, 4, 5,20   )  # This is a tuple with six elements
# Attempting to change an element will raise a TypeError
try:
    my_tuple[0] = 10  # This will raise an error
except TypeError as e:  
    print("Error:", e)  # Output: Error: 'tuple' object does not support item assignment    
```

```python
## Tuple Methods
# Count Method
my_tuple = (1, 2, 3, 2, 4, 2)
count_2 = my_tuple.count(2)
print("Count of 2 in my_tuple:", count_2)  # Output: 3 (number of times 2 appears in my_tuple)
# Index Method
index_3 = my_tuple.index(3)
print("Index of 3 in my_tuple:", index_3)  # Output: 2 (index of the first occurrence of 3 in my_tuple)     
```

```python
## Packing and Unpacking tuple
## packing when we create a tuple without using parentheses, it is called packing. The values are packed into a tuple automatically.
packed_tuple = 1, "hello", 3.14  # This is a packed tuple
print("Packed Tuple:", packed_tuple)  # Output: (1, 'hello', 3.14)
## unpacking is the process of assigning the elements of a tuple to individual variables. This is done by using the same number of variables as there are elements in the tuple.
a, b, c = packed_tuple  # Unpacking the tuple into individual variables
print("Unpacked values:", a, b, c)  # Output: Unpacked values: 1 hello 3.14
print(a)
print(b)
```

```python
## Unpacking with *
# Unpacking with * allows you to unpack a tuple into variables while collecting the remaining elements into a list.
my_tuple = (1, 2, 3, 4, 5)
a, *b, c = my_tuple  # Unpacking with * to collect the middle elements into a list
print("a:", a)  # Output: 1 (first element of the tuple)
print("b:", b)  # Output: [2, 3, 4] (list of the middle elements)
print("c:", c)  # Output: 5 (last element of the tuple)
```

```python
## Nested Tuple
## Nested Lists are lists that contain other lists as elements. Similarly, nested tuples are tuples that contain other tuples as elements. Nested tuples can be useful for representing hierarchical data or grouping related information together.
nested_tuple = (1, 2, (3, 4), (5, 6),[3,4,6])  # This is a nested tuple containing two inner tuples
print("Nested Tuple:", nested_tuple)  # Output: (1, 2, (3, 4), (5, 6), [3, 4, 6])
# Accessing elements in a nested tuple
print("First element:", nested_tuple[0])  # Output: 1 (first element of the outer tuple)
print("Second inner tuple:", nested_tuple[3])  # Output: (5, 6) (second inner tuple)
print("Third element (list):", nested_tuple[4])  # Output: [3, 4, 6] (third element, which is a list)
# Accessing elements within the inner tuples
print("First element of the first inner tuple:", nested_tuple[2][0])  # Output: 3 (first element of the first inner tuple)
print("Second element of the second inner tuple:", nested_tuple[3][1])  # Output: 6 (second element of the second inner tuple)
print("First element of the third element (list):", nested_tuple[4][0])  # Output: 3 (first element of the third element, which is a list)
```

```python
## iterating over nested tuples
print(nested_tuple)  # Output: (1, 2, (3, 4), (5, 6), [3, 4, 6])
for item in nested_tuple:
    if isinstance(item, tuple):  # Check if the item is a tuple
        print("Inner Tuple:", item)  # Output: Inner Tuple: (3, 4) and Inner Tuple: (5, 6)
    else:
        print("Element:", item)  # Output: Element: 1, Element: 2, and Element: [3, 4, 6]   
```
