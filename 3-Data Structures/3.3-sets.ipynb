#### Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
##create a set
my_set = {1, 2, 3, 4, 5}
print(my_set)  # Output: {1, 2, 3, 4, 5}
print(type(my_set))  # Output: <class 'set'>

my_set=set([1,2,3,4,5,6])
print(my_set)  # Output: {1, 2, 3, 4, 5, 6}
```

```python
my_empty_set=set([1,2,3,6,5,4,5,6])
print(my_empty_set)
```

```python
## Basics Sets Operation
## Adding and Removing Elements
## clear() method to remove all elements from the set
my_set.add(7)
print(my_set)  # Output: {1, 2, 3, 4, 5, 6, 7}  to add an element to the set
my_set.remove(3)
print(my_set)  # Output: {1, 2, 4, 5, 6, 7}  to remove an element from the set
my_set.discard(10)  # Does not raise an error if the element is not present
print(my_set)  # Output: {1, 2, 4, 5, 6, 7}
## pop method to remove and return an arbitrary element from the set
removed_element = my_set.pop()
print(f"Removed element: {removed_element}")  # Output: Removed element: 1
print(my_set)  # Output: {2, 4, 5, 6, 7} to check the set after popping an element
my_set.clear()  # to remove all elements from the set
print(my_set)  # Output: set() to check if the set is empty after clearing
```

```python
## MAthematical Operation
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
union_set = set_a.union(set_b)
print(union_set)  # Output: {1, 2, 3, 4, 5, 6} to find the union of two sets
intersection_set = set_a.intersection(set_b)
print(intersection_set)  # Output: {3, 4} to find the intersection of two sets
difference_set = set_a.difference(set_b)
print(difference_set)  # Output: {1, 2} to find the difference between two sets
symmetric_difference_set = set_a.symmetric_difference(set_b)
print(symmetric_difference_set)  # Output: {1, 2, 5, 6} to find the symmetric difference between two sets
```

```python
## differnces 
set1={1,2,3,4,5}
set2={4,5,6,7,8,9}  
print(set1.difference(set2))  # Output: {1, 2, 3}
## Symmetric Difference
set1.symmetric_difference(set2)  # Output: {1, 2, 3, 6, 7, 8, 9}
```

```python
## Sets Methods
set1={1,2,3,4,5}
set2={3,4,5}

## is subset
print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
### Counting Unique words in text

text="In this tutorial we are discussing about sets"
words=text.split()

## convert list of words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```

#### Conclusion
Sets are a powerful and flexible data type in Python that provide a way to store collections of unique elements. They support various operations such as union, intersection, difference, and symmetric difference, which are useful for mathematical computations. Understanding how to use sets and their associated methods can help you write more efficient and clean Python code, especially when dealing with unique collections and membership tests.
