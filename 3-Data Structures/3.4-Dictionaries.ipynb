##### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs.
Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## Creating Dictionaries
my_dict = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
print(my_dict)  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
print(type(my_dict))  # Output: <class 'dict'>
```

```python
## Accessing Dictionary elements
employee = {
    "name": "John Doe",
    "age": 28,
    "department": "Sales"
}
print(employee["name"])  # Output: John Doe 
print(employee["age"])  # Output: 28
print(employee["department"])  # Output: Sales

## Accessing using get() method
print(employee.get("name"))  # Output: John Doe
print(employee.get("age"))  # Output: 28
print(employee.get("department"))  # Output: Sales  
```

```python
## Dictionary methods
employee = {
    "name": "John Doe",
    "age": 28,
    "department": "Sales"
}
print(employee.keys())  # Output: dict_keys(['name', 'age', 'department'])
print(employee.values())  # Output: dict_values(['John Doe', 28, 'Sales'])
print(employee.items())  # Output: dict_items([('name', 'John Doe'), ('age', 28), ('department', 'Sales')]) 
```

```python
## shallow copy
original_dict = {"name": "Alice", "age": 30}
shallow_copy = original_dict.copy()
print(shallow_copy)  # Output: {'name': 'Alice', 'age': 30}
```

```python
### Iterating Over Dictionaries
for key in employee:
    print(key)  # Output: name, age, department

for value in employee.values():
    print(value)  # Output: John Doe, 28, Sales

for key, value in employee.items():
    print(f"{key}: {value}")  # Output: name: John Doe, age: 28, department: Sales
```

```python
## Nested Disctionaries
nested_dict = {
    "employee1": {
        "name": "Alice",
        "age": 30,
        "department": "HR"
    },
    "employee2": {
        "name": "Bob",
        "age": 25,
        "department": "IT"
    }
}
print(nested_dict)  # Output: {'employee1': {'name': 'Alice', 'age': 30, 'department': 'HR'}, 'employee2': {'name': 'Bob', 'age': 25, 'department': 'IT'}}
print(nested_dict["employee1"]["name"])  # Output: Alice
print(nested_dict["employee2"]["department"])  # Output: IT
print(nested_dict.items())  # Output: dict_items([('employee1', {'name': 'Alice', 'age': 30, 'department': 'HR'}), ('employee2', {'name': 'Bob', 'age': 25, 'department': 'IT'})])
```

```python
## Iterating over nested dictionaries
for employee_key, employee_info in nested_dict.items():
    print(f"{employee_key}:")
    for key, value in employee_info.items():
        print(f"  {key}: {value}")

## Dictionary Comprehension
squared_dict = {x: x**2 for x in range(1, 6)}
print(squared_dict)  # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
## Condition dictionary comprehension
even_squared_dict = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squared_dict)  # Output: {2: 4, 4: 16, 6: 36, 8: 64, 10: 100}
```

```python
## Practical Examples

## USe a dictionary to count he frequency of elements in list
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 dictionaries into one
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)  # Output: {'a': 1, 'b': 2, 'c': 3, 'd': 4}  
```
