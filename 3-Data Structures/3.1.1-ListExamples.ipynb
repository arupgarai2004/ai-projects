### Real-World Examples Using Lists in Python
Lists are one of the most commonly used data structures in Python, thanks to their versatility and ease of use. Here are several practical examples that illustrate their use in real-world scenarios

Example 1. Manage A To Do List
- Create a To Do List To Keep Track OF Tasks

```python
## Create To Do List  to keep track of tasks 
to_do_list = ["Buy groceries", "Clean the house", "Pay bills", "Call mom", "Finish AI project"]
## Add a new task to the list
to_do_list.append("Go for a walk")
print(to_do_list)  # Output: ['Buy groceries', 'Clean the house', 'Pay bills', 'Call mom', 'Finish AI project', 'Go for a walk']
## Remove a task from the list
to_do_list.remove("Pay bills")
print(to_do_list)  # Output: ['Buy groceries', 'Clean the house', 'Call mom', 'Finish AI project', 'Go for a walk']
## Check if a task is in the list
if "Call mom" in to_do_list:
    print("Don't forget to call mom!")  # Output: Don't forget to call mom! 

## print remaining tasks using a loop
print("Remaining tasks:")
for task in to_do_list:
    print(f"- {task}")
    
```

#### Example 2: Organizing Student Grades
- Create a list to store and calculate average grades for students

```python
# Organizing student grades
student_grades = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 90
}
## add a new student and grade
student_grades["Eve"] = 88
print(student_grades)  # Output: {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'David': 90, 'Eve': 88}
## Calculate the average grade
average_grade = sum(student_grades.values()) / len(student_grades)
print(f"Average grade: {average_grade:.2f}")  # Output: Average grade: 86.60
## Find the student with the highest grade and find the student with the lowest grade
top_student = max(student_grades, key=student_grades.get)
lowest_student = min(student_grades, key=student_grades.get)
print(f"Top student: {top_student} with grade {student_grades[top_student]}")  # Output: Top student: Bob with grade 92 
print(f"Lowest student: {lowest_student} with grade {student_grades[lowest_student]}")  # Output: Lowest student: Charlie with grade 78
```

#### Example 3: Managing An Inventory
- Use a list to manage inventory items in a store

```python
# Managing an inventory
inventory = {
    "apples": 50,
    "bananas": 30,
    "oranges": 25
}
## Add more items to the inventory
inventory["grapes"] = 40
inventory["strawberries"] = 20
print(inventory)  # Output: {'apples': 50, 'bananas': 30, 'oranges': 25, 'grapes': 40, 'strawberries': 20}
## Update the quantity of an item
inventory["apples"] = 60
print(inventory)  # Output: {'apples': 60, 'bananas': 30, 'oranges': 25, 'grapes': 40, 'strawberries': 20}
## Remove an item from the inventory
del inventory["oranges"]
print(inventory)  # Output: {'apples': 60, 'bananas': 30, 'grapes': 40, 'strawberries': 20}
## Check if an item is in the inventory and print its quantity if its less than 1 says "out of stock"
if "bananas" in inventory:
    if inventory["bananas"] < 1:
        print("Bananas are out of stock!")
    else:
        print(f"Bananas are in stock! Quantity: {inventory['bananas']}")  # Output: Bananas are in stock! Quantity: 30
```

#### Example 4:Collecting User Feedback
- Use a list to collect and analyze user feedback.

```python
# colecting user Feedback
user_feedback = {
    "Alex": "Great product! Highly recommend.",
    "Alice": "Good value for the price.",
    "Bob": "Not satisfied with the quality.",
    "Charlie": "Excellent customer service."
}
## Add new feedback
user_feedback["David"] = "Fast shipping and great packaging."
print(user_feedback)  # Output: {'Alex': 'Great product! Highly recommend.', 'Alice': 'Good value for the price.', 'Bob': 'Not satisfied with the quality.', 'Charlie': 'Excellent customer service.', 'David': 'Fast shipping and great packaging.'}
# count positive and negative feedback
positive_count = sum(1 for feedback in user_feedback.values() if "great" in feedback.lower() or "excellent" in feedback.lower() or "recommend" in feedback.lower())
negative_count = sum(1 for feedback in user_feedback.values() if "not satisfied" in feedback.lower() or "bad" in feedback.lower() or "poor" in feedback.lower())
print(f"Positive feedback count: {positive_count}")  # Output: Positive feedback count: 3
print(f"Negative feedback count: {negative_count}")  # Output: Negative feedback count: 1
```
