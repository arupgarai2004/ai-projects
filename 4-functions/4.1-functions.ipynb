Introduction to Functions

Definition:

A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability.

```python
# function define, calling , function parameters, variable length argument and return statement examples

def greet(name): # function definition with parameter name
    return f"Hello, {name}!" # return statement to return a greeting message

print(greet("Alice"))  # Output: Hello, Alice!  # function call with argument "Alice"   
def add(a, b): # function definition with parameters a and b
    return a + b  # return statement to return the sum of a and b
print(add(5, 3))  # Output: 8  # function call with arguments 5 and 3       
def variable_length_args(*args): # function definition with variable length arguments
    return sum(args)  # return statement to return the sum of all arguments
print(variable_length_args(1, 2, 3, 4))  # Output: 10  # function call with multiple arguments  
# function with default parameter value
def greet(name="Guest"): # function definition with default parameter value "Guest"
    return f"Hello, {name}!"  # return statement to return a greeting message
print(greet())  # Output: Hello, Guest!  # function call without argument, uses default value
print(greet("Bob"))  # Output: Hello, Bob!  # function call with argument "Bob", overrides default value    
# function with keyword arguments
def greet(name, greeting="Hello"): # function definition with parameters name and greeting with default value "Hello"
    return f"{greeting}, {name}!"  # return statement to return a greeting message
print(greet(name="Charlie", greeting="Hi"))  # Output: Hi, Charlie!  # function call with keyword arguments, overrides default value
# returning multiple values from a function
def calculate(a, b): # function definition with parameters a and b
    sum_result = a + b  # calculate the sum of a and b
    product_result = a * b  # calculate the product of a and b
    return sum_result, product_result  # return both results as a tuple
sum_value, product_value = calculate(4, 5)  # function call with arguments 4 and 5, unpacking the returned tuple into sum_value and product_value
print(f"Sum: {sum_value}, Product: {product_value}")  # Output: Sum: 9, Product: 20
```
