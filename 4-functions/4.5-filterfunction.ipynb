The filter() Function in Python

The filter() function constructs an iterator from elements of an iterable for which a function returns true. It is used to filter out items from a list (or any other iterable) based on a condition.


```python
## filter function example
def even(num):
    return num % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(even, numbers))
print(even_numbers)  # Output: [2, 4, 6]

## Using lambda function with filter
even_numbers_lambda = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers_lambda)  # Output: [2, 4, 6]
     

```

```python
## Filter() to check  the person come from a specific city, city will be passed as an argument to the function
def from_city(person, city):
    return person["city"] == city   
people = [
    {"name": "Alice", "city": "New York"},
    {"name": "Bob", "city": "Los Angeles"},
    {"name": "Charlie", "city": "New York"},
    {"name": "David", "city": "Chicago"}
]
new_york_people = list(filter(lambda person: from_city(person, "New York"), people))
print(new_york_people)  # Output: [{'name': 'Alice', 'city': 'New York'}, {'name': 'Charlie', 'city': 'New York'}]  
# using lambda function directly without defining a separate function
new_york_people_lambda = list(filter(lambda person: person["city"] == "Chicago", people))
print(new_york_people_lambda)  # Output: [{'name': 'David', 'city': 'Chicago'}]
```
